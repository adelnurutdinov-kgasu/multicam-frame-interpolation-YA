# Ансамбль: Consensus U-Net (front/rear) + RIFE

Перебор коэффициентов смешивания предсказания модели и RIFE на **val**-сплите
(тот же `seed=42`, `val_fraction=0.16`, что в `consensus_training.ipynb`).

**Базовые режимы:**
- `linear`: `α·model + (1-α)·rife`
- `cov_soft`: `cov^γ·model + (1-cov^γ)·rife`
- `cov_gate`: где `cov ≥ thr` → model, иначе RIFE
- `alpha_cov`: `w = α + β·cov`, затем `w·model + (1-w)·rife`
- `conf_gate`: по `confidence` голове модели (если есть в чекпоинте)


```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch

REPO = Path(r"C:/Users/adel/Documents/GitHub/YA-")
if str(REPO) not in sys.path:
    sys.path.insert(0, str(REPO))

from consensus_kit import (
    ConsensusConfig,
    blend_alpha_cov,
    blend_conf_gate,
    blend_cov_gate,
    blend_cov_soft,
    blend_linear,
    default_config,
    discover_samples,
    eval_cache_psnr,
    load_consensus_model,
    make_split,
    psnr_uint8_np,
    run_inference_cache,
)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(DEVICE)
```

```python
# --- пути (не перезаписывает side-чекпоинты) ---
CFG = default_config()
CFG.allowed_cameras = ("front", "rear")

CKPT_PATH = REPO / "checkpoints_consensus" / "consensus_unet_best.pt"
USE_EMA = True
CACHE_NPY = REPO / "checkpoints_consensus" / "val_cache_front_rear.npz"  # опционально
REBUILD_CACHE = True  # False — загрузить кэш с диска

VAL_SUBSET = 0  # 0 = весь val; иначе первые N для быстрого прогона

print(f"ckpt: {CKPT_PATH}")
print(f"exists: {CKPT_PATH.is_file()}")
```

```python
all_samples = discover_samples(CFG)
_, val_paths = make_split(all_samples, CFG.val_fraction, CFG.seed)
if VAL_SUBSET > 0:
    val_paths = val_paths[:VAL_SUBSET]
print(f"val samples: {len(val_paths)} / {len(all_samples)} total")

if not CKPT_PATH.is_file():
    raise FileNotFoundError(f"Нет чекпоинта: {CKPT_PATH}\nСначала обучи consensus_training.ipynb")

model, ckpt_meta = load_consensus_model(CKPT_PATH, DEVICE, use_ema=USE_EMA, cfg=CFG)
vp = ckpt_meta.get("val_psnr_ema", ckpt_meta.get("val_psnr"))
print(f"loaded  val_psnr (from ckpt): {vp}")
```

```python
def _save_cache(cache, path):
    # список dict → npz со стеками (для ускорения повторных прогонов)
    keys = ["model", "rife", "gt", "cov", "base"]
    stacks = {k: np.stack([c[k] for c in cache], axis=0) for k in keys}
    if cache and "conf" in cache[0]:
        stacks["conf"] = np.stack([c["conf"] for c in cache], axis=0)
    meta_json = [c["meta"] for c in cache]
    import json
    path.parent.mkdir(parents=True, exist_ok=True)
    np.savez_compressed(path, **stacks, meta_json=np.array([json.dumps(m) for m in meta_json], dtype=object))
    print(f"saved cache → {path}")


def _load_cache(path):
    import json
    z = np.load(path, allow_pickle=True)
    n = z["model"].shape[0]
    cache = []
    for i in range(n):
        e = {k: z[k][i] for k in ("model", "rife", "gt", "cov", "base")}
        e["meta"] = json.loads(str(z["meta_json"][i]))
        if "conf" in z.files:
            e["conf"] = z["conf"][i]
        cache.append(e)
    print(f"loaded cache ← {path}  ({n} samples)")
    return cache


if REBUILD_CACHE or not CACHE_NPY.is_file():
    cache = run_inference_cache(model, val_paths, CFG, DEVICE)
    _save_cache(cache, CACHE_NPY.with_suffix(".npz"))
else:
    cache = _load_cache(CACHE_NPY.with_suffix(".npz"))

# baselines
b_model = eval_cache_psnr(cache, lambda e: e["model"])
b_rife = eval_cache_psnr(cache, lambda e: e["rife"])
b_base = eval_cache_psnr(cache, lambda e: e["base"])
print(f"model only: {b_model:.3f} dB")
print(f"RIFE only:  {b_rife:.3f} dB")
print(f"base_init:  {b_base:.3f} dB")
```

## 1. Линейный ансамбль `α·model + (1-α)·rife`

```python
alphas = np.linspace(0, 1, 21)
rows = []
for a in alphas:
    p = eval_cache_psnr(cache, lambda e, a=a: blend_linear(e["model"], e["rife"], a))
    rows.append({"alpha": a, "psnr": p, "d_vs_rife": p - b_rife, "d_vs_model": p - b_model})
df_linear = pd.DataFrame(rows)
best_lin = df_linear.loc[df_linear["psnr"].idxmax()]
print(df_linear.sort_values("psnr", ascending=False).head(8).to_string(index=False))
print(f"\nBEST linear α={best_lin['alpha']:.2f}  PSNR={best_lin['psnr']:.3f} dB")

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(df_linear["alpha"], df_linear["psnr"], "o-", label="ensemble")
ax.axhline(b_model, color="C1", ls="--", label=f"model ({b_model:.2f})")
ax.axhline(b_rife, color="C2", ls="--", label=f"RIFE ({b_rife:.2f})")
ax.set_xlabel("α (вес модели)")
ax.set_ylabel("PSNR dB")
ax.set_title("Linear blend on val")
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 2. Coverage: мягкое смешивание и порог

Идея: в зонах с хорошим warp (высокий `coverage`) больше доверять модели, в дырках — RIFE.

```python
# cov_soft: gamma
gammas = [0.5, 1.0, 1.5, 2.0, 3.0]
for g in gammas:
    p = eval_cache_psnr(cache, lambda e, g=g: blend_cov_soft(e["model"], e["rife"], e["cov"], g))
    print(f"cov_soft γ={g:.1f}  PSNR={p:.3f}  ΔRIFE={p-b_rife:+.3f}")

print()
for thr in np.linspace(0.1, 0.9, 9):
    p = eval_cache_psnr(cache, lambda e, t=thr: blend_cov_gate(e["model"], e["rife"], e["cov"], t))
    print(f"cov_gate thr={thr:.2f}  PSNR={p:.3f}  ΔRIFE={p-b_rife:+.3f}")
```

## 3. 2D сетка `α + β·coverage`

```python
alphas_1d = np.linspace(0, 0.6, 13)
betas = np.linspace(0, 1.2, 13)
grid = np.zeros((len(betas), len(alphas_1d)))
for i, b in enumerate(betas):
    for j, a in enumerate(alphas_1d):
        grid[i, j] = eval_cache_psnr(
            cache,
            lambda e, a=a, b=b: blend_alpha_cov(e["model"], e["rife"], e["cov"], a, b),
        )
ij = np.unravel_index(grid.argmax(), grid.shape)
print(f"BEST α={alphas_1d[ij[1]]:.2f} β={betas[ij[0]]:.2f}  PSNR={grid[ij]:.3f} dB")

fig, ax = plt.subplots(figsize=(9, 6))
im = ax.imshow(grid, aspect="auto", origin="lower", cmap="viridis",
               extent=[alphas_1d[0], alphas_1d[-1], betas[0], betas[-1]])
plt.colorbar(im, ax=ax, label="PSNR dB")
ax.set_xlabel("α (базовый вес модели)")
ax.set_ylabel("β (усиление от coverage)")
ax.set_title("α + β·cov blend")
plt.tight_layout()
plt.show()
```

## 4. Confidence gate (если есть в чекпоинте)

```python
has_conf = "conf" in cache[0]
print(f"confidence map in cache: {has_conf}")
if has_conf:
    for thr in np.linspace(0.3, 0.9, 13):
        p = eval_cache_psnr(cache, lambda e, t=thr: blend_conf_gate(e["model"], e["rife"], e["conf"], t))
        print(f"conf_gate thr={thr:.2f}  PSNR={p:.3f}  ΔRIFE={p-b_rife:+.3f}")
else:
    print("Пропуск — модель не сохранила confidence в кэш (пересобери REBUILD_CACHE=True)")
```

## 5. Визуализация лучшего линейного α

Подставь свои лучшие параметры из сеток выше.

```python
BEST_ALPHA = float(best_lin["alpha"])  # или вручную, напр. 0.65
BEST_BLEND = lambda e: blend_linear(e["model"], e["rife"], BEST_ALPHA)


def show_compare(entry, title_extra=""):
    gt, m, r = entry["gt"], entry["model"], entry["rife"]
    ens = BEST_BLEND(entry)
    meta = entry["meta"]
    err_m = np.abs(m - gt).mean(axis=2)
    err_r = np.abs(r - gt).mean(axis=2)
    err_e = np.abs(ens - gt).mean(axis=2)
    win = err_r > err_e + 0.01  # зелёный = ансамбль лучше RIFE
    fig, axes = plt.subplots(2, 4, figsize=(16, 7))
    fig.suptitle(f"{meta['sample_id'][:50]}… | {meta['camera']} {title_extra}", fontsize=10)
    for ax, img, t in zip(
        axes[0],
        [gt, r, m, ens],
        ["GT", f"RIFE {psnr_uint8_np(r,gt):.2f}", f"model {psnr_uint8_np(m,gt):.2f}", f"ens α={BEST_ALPHA:.2f} {psnr_uint8_np(ens,gt):.2f}"],
    ):
        ax.imshow(np.clip(img, 0, 1))
        ax.set_title(t, fontsize=9)
        ax.axis("off")
    for ax, err, t in zip(axes[1, :3], [err_r, err_m, err_e], ["|RIFE-GT|", "|model-GT|", "|ens-GT|"]):
        ax.imshow(err, cmap="hot", vmin=0, vmax=0.25)
        ax.set_title(t, fontsize=9)
        ax.axis("off")
    wm = np.full(err_e.shape + (3,), 0.5)
    wm[win] = [0.15, 0.85, 0.15]
    axes[1, 3].imshow(wm)
    axes[1, 3].set_title(f"ens<RIFE {100*win.mean():.1f}% px", fontsize=9)
    axes[1, 3].axis("off")
    plt.tight_layout()
    plt.show()


# лучший и худший по PSNR ансамбля
scored = [(i, psnr_uint8_np(BEST_BLEND(e), e["gt"])) for i, e in enumerate(cache)]
scored.sort(key=lambda x: x[1])
print("worst:", scored[0][1], "best:", scored[-1][1])
for idx, ps in [scored[0], scored[len(scored)//2], scored[-1]]:
    show_compare(cache[idx], f"PSNR={ps:.2f}")
```
