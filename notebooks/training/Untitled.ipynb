```python
'''А) Написать программу, которая на вход получает натуральное число N и
выводит N-е число ряда Фибоначчи на экран.
Пояснение: последовательность Фибоначчи — это ряд чисел, в котором
первые два элемента равны 1, а каждый следующий элемент равен сумме двух
предыдущих: 1, 1, 2, 3, 5, 8, 13, 21, ...'''

n = int(input())
a = 1
b = 1
if n >= 3:
    for i in range(n-2):
        a_new = a + b
        b = a
        a = a_new
print(a)
```

```python
'''В) Числа последовательно вводятся в цикле. Ввод завершается числом 0.
Определите индекс наибольшего элемента последовательности. Если
наибольших элементов несколько, выведите индекс первого из них. Нумерация
элементов начинается с нуля.
'''
max_n = None
temp = None
max_ind = None
counter = -1

while True:
    temp = float(input())
    if temp == 0:
        break
    counter += 1 
    if max_n == None or temp > max_n:
        max_n = temp
        max_ind = counter
print(f"Наибольший элемент: {max_n}\nИндекс первого: {max_ind} ")
```

```python
'''В) Числа последовательно вводятся в цикле. Ввод завершается числом 0.
Определите индекс наибольшего элемента последовательности. Если
наибольших элементов несколько, выведите индекс первого из них. Нумерация
элементов начинается с нуля.
'''
max_n = None
temp = None
max_ind = None
counter = -1

while temp != 0:
    temp = float(input())
    counter += 1 
    if max_n == None or temp > max_n:
        max_n = temp
        max_ind = counter
print(f"Наибольший элемент: {max_n}\nИндекс первого: {max_ind} ")
```

```python
'''Напишите программу, которая принимает на вход список букв в одной
строке (через генератор). Сформируйте список из гласных букв'''

n = str(input())
vowels = set('eyuioaEYUIOAуеыаоэяиюёУЕЫАОЭЯИЮЁ')

list_let = [let for let in n if let in vowels and let.isalpha()]
print(list_let)
```

```python
'''А) Напишите программу, которая анализирует текст на основе словаря,
содержащий название города и количество жителей и сообщает к какой
категории этот город относится.
По признаку численности населения города классифицируются следующим
образом:
 свыше 1 млн чел – крупнейшие;
 от 250 тыс чел – до 1 млн чел – крупные;
 от 100 до 250 тыс чел – большие;
 от 50 до 100 тыс чел – средние;
 до 50 тыс чел – малые.'''

city_dict = {
    "москва": 13149803,
    "санкт-Петербург": 5597763,
    "новосибирск": 1633851,
    "екатеринбург": 1536183,
    "казань": 1318604,
    "нижний Новгород": 1204985,
    "челябинск": 1177058,
    "уфа": 1163304,
    "самара": 1158952,
    "ростов-на-Дону": 1140487,
    "краснодар": 1138654,
    "пермь": 1026908,
    "волгоград": 1018898,
    "саратов": 887365,
    "тюмень": 861098,
    "плёс": 1896,
    "мышкин": 2230,
    "касимов": 23500,
    "тотьма": 8647,
    "сольвычегодск": 1952
}

corpus = input().split()
for word in corpus:
    if word.lower() in city_dict:
        if city_dict[word] >= 1000000:
            result = 'крупнейший'
        elif city_dict[word] >= 250000:
            result = 'крупный'
        elif city_dict[word] >= 100000:
            result = 'большой'
        elif city_dict[word] >= 50000:
            result = 'средний'
        else:
            result = 'малый'
        print(word, result)
```

```python
'''до 50 тыс чел – малые.
В) Выполните на основе набора данных data.json задание: Какой клиент
(client_id) совершил наибольшее количество действий с (category) = datepicker'''

import json
from collections import Counter
with open('data.json', mode = 'r', encoding = 'cp1251') as f:
    data = json.load(f)
    
data = data.get('events_data', []) 
target_category = 'datepicker'

ids = [item['id'] for item in data if item.get('category') == target_category]
max_count = Counter(ids).most_common(1)
print(f"Клиент с максимальным количеством действий с {target_category}: {max_count[0][0]}")
print(f"Kоличество действий с {target_category}: {max_count[0][1]}")
```
